```python
#input_file_path = "aoc_2025_11_input_sample.txt"
input_file_path = "aoc_2025_11_input.txt"

content = None
with open(input_file_path, "r") as file:
    content = file.readlines()
```

```python
puzzle = {}
for line in content:
    parts = line.split(" ")
    puzzle[parts[0][:-1]] = [x.replace("\n", "") if "\n" in x else x for x in parts[1:]]
    
```

```python
def dfs(
    start: str, end: str, graph: dict
) -> int:
    visited = set()
    stack = [start]
    paths = 0

    while len(stack) > 0:
        node = stack.pop()
        
        if node == end:
            paths += 1
            continue
            
        if node not in visited:
            visited.add(node)
            
        stack += graph[node]

    return paths
```

```python
print(dfs("you", "out", puzzle))
```
