```python
input_file_path = "aoc_2025_11_input_sample2.txt"
#input_file_path = "aoc_2025_11_input.txt"

content = None
with open(input_file_path, "r") as file:
    content = file.readlines()
```

```python
puzzle = {}
for line in content:
    parts = line.split(" ")
    puzzle[parts[0][:-1]] = [x.replace("\n", "") if "\n" in x else x for x in parts[1:]]
```

```python
def dfs(
    start: str, end: str, graph: dict
) -> int:
    stack = [[start, [start]]]
    paths = 0

    while stack:
        node = stack.pop()
        
        if node[0] == end and "fft" in node[1] and "dac" in node[1]:
            paths += 1
            continue
            
        try:
            for n in graph[node[0]]:
                new_node = [n, node[1]+[n]]
                stack.append(new_node)
        except KeyError:
            pass

    return paths
```

```python
dfs("svr", "out", puzzle)
```
