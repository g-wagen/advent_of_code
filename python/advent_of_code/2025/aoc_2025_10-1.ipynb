```python
def toggle(light) -> int:
    mapping = {0: 1, 1: 0}
    return mapping[light]
```

```python
def push_button(
    lights: list[int], 
    sequence: list[int]
) -> list[int]:
    out = [x for x in lights]
    for i in sequence:
        out[i] = toggle(out[i]) 
    return out
```

```python
def bfs(
    start: list[int], 
    target: list[int], 
    buttons: list[tuple[int]],
) -> int:
    """Breadth first search.
    Should stop when start equals target.
    Returns button press count when stopped"""

    button_presses = 0
    queue = [b for b in buttons]
    current = [x for x in start]
    
    while current != target:
        for button in buttons:
            currentcopy = [c for c in current]
            currentcopy = push_button(currentcopy, button)
        
            current = [c for c in currentcopy]
        button_presses += 1

    return button_presses
```

```python
from itertools import combinations_with_replacement

buttonpresses = []

for line in content:
    current_butts = []
    line_content = line.split(" ")
    #joltage = line_content[-1][1:-2]
    
    starting_lights = [0] * len(line_content[0][1:-1])
    target_lights = [0 if x == "." else 1 for x in line_content[0][1:-1]]
    
    buttons = [list(int(y) for y in x[1:-1].split(",")) for x in line_content[1:-1]]
    
    combinations = list()

    for b, butt in enumerate(buttons):
        combinations += list(
                combinations_with_replacement(
                    buttons, 
                    b
                )
            )
        
    
    for comb in combinations:
        comb = list(comb)
        butt = 0
        current = [s for s in starting_lights]
        
        for item in comb:
            if len(item) < 1:
                continue
            current = push_button(current, item)
            butt += 1
            if current == target_lights:
                current_butts.append(butt)
                break
                
    min_press = min(current_butts)
    print(min_press)
    
    buttonpresses.append(min_press)
    

print(sum(buttonpresses))
    
    
```
